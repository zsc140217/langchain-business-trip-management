```python
# 安装依赖
!pip install -q transformers peft accelerate bitsandbytes

# 加载模型
from transformers import AutoModelForCausalLM, AutoTokenizer
from peft import LoraConfig, get_peft_model

model = AutoModelForCausalLM.from_pretrained('Qwen/Qwen2-7B-Instruct', load_in_4bit=True, device_map='auto')
tokenizer = AutoTokenizer.from_pretrained('Qwen/Qwen2-7B-Instruct')

# 配置LoRA
lora_config = LoraConfig(r=16, lora_alpha=32, target_modules=['q_proj','v_proj'], lora_dropout=0.05, task_type='CAUSAL_LM')
model = get_peft_model(model, lora_config)
model.print_trainable_parameters()

print('✅ 配置完成！参数量0.06%，类似YOLOv8只训练Head')
```
